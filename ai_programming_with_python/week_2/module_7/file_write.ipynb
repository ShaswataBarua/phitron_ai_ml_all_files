```python
#create a file then write

with open("./sample_data/text.txt","w") as file:
  file.write("My name is shaswata barua\n")
  file.write("This is file creating with write")


```

```python
#overwrite a file

with open("./sample_data/text.txt","w") as file:
  file.write("2nd attempt\n")
  file.write("This is file overwrite with write\n")

```

```python
# no overwrite just append

with open("./sample_data/text.txt","a") as file:
  file.write("\n My name is shaswata barua\n")
  file.write("This is file creating with write")
```
