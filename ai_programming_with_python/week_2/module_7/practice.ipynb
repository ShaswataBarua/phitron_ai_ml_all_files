Practice 1 :
Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt **file**

```python
from functools import  reduce


with open("./sample_data/test.txt","r") as file:
  stl = file.readlines()
  print(stl)
  #line ber kora
  print(len(stl))
  tl=len(stl)

  #word ber kor koto gula

  word_count = list(map(lambda x : len(x.split()),stl))

  print(word_count)

  #total word

  tw = reduce(lambda x,y : x+y,word_count)
  print(tw)

#char count

#1.line delete
string_list=list(map(str.strip,stl))
print(string_list)

#2.space delete
string_list =list(map(lambda x : x.replace(" ",""),string_list))

print(string_list)

#word count

string_list = list(map(lambda x : len(x),string_list  ))

print(string_list)

tc= reduce(lambda x,y : x+y, string_list)
print(tc)
```

```python

with open("./sample_data/counter_of_string.txt","w") as file:
  file.write(f" total lines :{tl}\n total number of words :{tw}\n total numbers of word :{tc}")
```
