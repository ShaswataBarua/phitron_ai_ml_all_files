DIFFERENT TYPE OF ARGUMENT

```python
def see(a,b,c):
  print(f"{a},{b},{c}")
```

```python
see(10,30,20)
```

```python
def see(*args):
  print(type(args),args)

see(10,20,30,40,50,60)
```

```python
def see(*args):
  for i in args :
    print(i)

see(10,20,30,40,50,60)
```

KEYWORD ARGUMENTS

```python
def see(a,b,c):
  print(f"{a},{b},{c}")

see(a=00,b=33,c=22)
```

```python
def adder(**kwargs):
  print(kwargs)

adder(a=10,b=30,c=40,n_estimated=999.98)
```

```python
def adder(**kwargs):
  for key,value in kwargs.items() :
    print(f"the keys are:{key} , the values are:{value}")

adder(a=10,b=30,c=40,n_estimated=999.98)
```
