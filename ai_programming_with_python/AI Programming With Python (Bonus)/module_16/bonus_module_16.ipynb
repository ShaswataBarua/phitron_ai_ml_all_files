## 16.10-1 ScatterPlot with plotly


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
students = pd.read_csv('students.csv') 
students
```

```python
student = pd.read_csv('sns_data.csv') 
student
```

```python
stu = pd.read_csv('student_dataset_complete.csv') 
stu
```

```python
student_marks = stu[['Marks','study_hours','attendance_rate','gender']]
student_marks
```

```python
fig = px.scatter( stu , x='study_hours' , y='Marks',
                color='gender',
                size='Tshirt_size',
                hover_data=['hostel']) 
fig.show()
```

## 16.10-2 Lineplot with plotly


```python
en = pd.read_csv('enrollment_data.csv') 
en
```

```python
fig = px.line( en , x='Year',y='Programming',markers=True) 
fig.show()
```

```python
fig.write_html('year_prog.html')
```

## 16.10-3 Histogram using plotly


```python
fig = px.histogram(student_marks, x='attendance_rate',color='gender',nbins=5)

fig.show()
```
