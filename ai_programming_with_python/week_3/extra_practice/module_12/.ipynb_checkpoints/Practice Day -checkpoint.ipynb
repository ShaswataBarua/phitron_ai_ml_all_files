## Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print students with Score > 60 and those with Score < 40.



```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan', 'Fiona', 'George', 'Hannah'],
    'Score': [85, 38, 92, 55, 78, 29, 65, 42]
}

df_students = pd.DataFrame(data)
df_students
```

```python
# Students with score > 60
high_scores = df_students.loc[df_students['Score'] > 60]

# Students with score < 40
low_scores = df_students.loc[df_students['Score'] < 40]

high_scores
```

```python
low_scores
```
