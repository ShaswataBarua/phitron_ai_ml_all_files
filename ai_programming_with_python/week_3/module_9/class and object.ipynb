```python
class phone :
  catagory = "Electronics"

  def __init__(self,model,battery,camera,battery_percentage=100):
    self.model=model
    self.battery=battery
    self.camera=camera
    self.battery_percentage=battery_percentage

  #method
  def charge(self,hour):
    print(f"charge completed by {hour} {self.model}")

  def capture(self,photo):
    if self.battery_percentage <= 0:
      print("NO CHARGE")
    else :
      self.battery_percentage -= photo
      print(f"PHOTO CAPTURED {self.model}")


class cooling_mechanism :
  def __init__(self,cooling_method):
    self.cooling_method = cooling_method

  #method
  def cooling_on(self):
    print(f"the system is being cool y {self.cooling_method}")
```

```python
#creating object

apple= phone()
apple.catagory= "super intelligent"
xioami=phone()
nokia = phone()
print(type(apple))
print(apple.catagory)
print(xioami.catagory)
print(nokia.catagory)
```

```python
apple=phone("17 promax",3000,45)
xioami=phone("note 7 pro",3500,45)
nokia=phone("b17",10000,20)

print(nokia.battery)
print(nokia.battery_percentage)

xioami.capture(20)
print(xioami.battery_percentage)
print(nokia.battery_percentage)
```

SINGLE INHERITANCE

```python
#child class

class smartphone(phone):
  def __init__(self,model,battery,camera,processor):
    phone.__init__(self,model,battery,camera) # Changed from super() to direct call
    self.processor = processor

  def charge(self, hour):
    print("SUPER CHARGE IN PROCESS")
    phone.charge(self, hour) # Changed from super() to direct call
```

```python
pro = smartphone("x",23233,99,"adriuno")

print(pro.model)
print(pro.battery)
pro.charge(10)

pro.capture(100)
```

MULTIPLE IHERITANCE

```python
class smartphone_cooligmode(smartphone, cooling_mechanism):
  def __init__(self, model, battery, camera, processor, cooling_method):
    smartphone.__init__(self, model, battery, camera, processor)
    cooling_mechanism.__init__(self, cooling_method)
```

```python
pro_cooling = smartphone_cooligmode("apple",10000,48,"xenos","liquid nitrogen")

print(pro_cooling.model)
print(pro_cooling.battery)
print(pro_cooling.camera)
print(pro_cooling.processor)
print(pro_cooling.cooling_on())
```

POLYMORPHISM


```python
class camera:
  def __init__(self,name):
   self.name = name
#method
  def capture(self):
    print("a photo is captured")

```

```python
class smart(camera):
  def __init__(self, name,resolution):
    super().__init__(name)
    self.resolution=resolution

    #method overriding
  def capture(self):
    print("photo is captured by phone")

class dslr(camera):
  def __init__(self, name,resolution):
    super().__init__(name)
    self.resolution=resolution
  def capture(self):
    print("photo is captured by dslr")



class drone(camera):
  def __init__(self, name,resolution):
    super().__init__(name)
    self.resolution=resolution

  def capture(self):
    print("photo is captured by drone")
```

```python
phone1= smart("phone",48)
DSLR =dslr("xenon",190)
DRONE = drone("cmake",300)

phone1.capture()
DSLR.capture()
DRONE.capture()
```
