```python
import numpy as np

x = np.array([10,20,30,40])
y =np.array([2,3,4,5])
#faster for vectorization
add = x+y
print(add)

sub = x-y
print(sub)

mul = x*y
print(mul)

div = x/y
print(div)

rem = x%y
print(rem)
```

```python
# mathematical function

# trigonometry

# sin , cos , tan takes values in radian

x = np.array([10,20,30,40])

sin_val = np.sin(x)
print(sin_val)

cos_val = np.cos(x)
print(cos_val)

deg_con = np.rad2deg(x)
print(deg_con)
```

```python
#log 
x = np.array([10,16,30,40])

base10log_val = np.log10(x)
print(base10log_val)

two_diye_bhag= np.log2(x)
print(two_diye_bhag)

sqrt = np.sqrt(x)
print(sqrt)
```

```python
#different

sum = np.sum(x)
print(sum)

cumul = np.cumulative_sum(x)
print(cumul)
```

Logical Function

```python
# comparison - >,< , >= ,<= , == , !=

x = np.array([10,20,30,40])
y =np.array([20,3,4,5])

gt = x>y
print(gt)

equal = x==y
print(equal)

# all () -> shob true kina
# any() -> jekono ekta true ase kina

print(np.all(gt))
print(np.any(equal))
```

```python
#inplace

x = np.array([10 , 8 , 30, 99])
z= x.copy()

print(z)

z.sort()
print(z)
```

```python
#copy sorting

print(x)

sort_arr= np.sort(x)

print(sort_arr)
print(x)
```

```python
#2d array sorting

mat = np.array([[10,23,9] , [9,1,4]])
print(mat)

hor_sort = np.sort(mat,axis=1)
print(hor_sort)

ver_sort = np.sort(mat,axis=0)
print(ver_sort)
```

```python
#np.where ( condition , x , y ) if true x , otherwise replace with y

index = np.where(x==8)
print(index)

index = np.where(x>8)
print(index)

arr= np.where(x>8,x,0)
print(arr)
```

counting in ndarray


```python
a = np.random.randint(1 ,100 , size =(100,))
print(a)
```

```python
val_gt_count = np.count_nonzero(a>10)
print(val_gt_count)
```

```python
unique_val = np.unique(a)
print(unique_val)
```

```python
unique_val = np.unique(a, return_counts=True)
print(unique_val)
```

```python
unique_val,count = np.unique(a, return_counts=True)
print(unique_val)
print(count)
```
