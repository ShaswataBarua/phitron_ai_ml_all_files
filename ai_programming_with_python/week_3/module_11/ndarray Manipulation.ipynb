```python
import numpy as np

arr = np.random.randint(1,100,size=(10,5))# sample - 10 , feature - 5

print(arr)
```

```python
print(arr.shape)
```

```python
b = arr.reshape(5,10)
print(b)
print(b.shape)
print(b.ndim)
```

```python
flatten = b.flatten()

print(flatten)
flatten.ndim
```

```python
colmun_flatten = np.ravel(b,order='F')

print(colmun_flatten)
```

```python

row_flatten = np.ravel(b,order='C')

print(row_flatten)
```

```python
a = np.random.randint(1,10,size=(2,3))
b = np.random.randint(20,30,size=(2,3))
print(a)
print(b)
```

concatenate

```python
con_row = np.concatenate((a,b),axis=0) # row wise

print(con_row)

con_col = np.concatenate((a,b),axis=1) # col wise

print(con_col)
```
