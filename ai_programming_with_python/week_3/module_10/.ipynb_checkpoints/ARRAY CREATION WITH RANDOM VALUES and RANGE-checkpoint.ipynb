ARRAY CREATION WITH RANDOM VALUES

```python
import numpy as np


arr = np.random.rand(2,3)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.random.randint(1,10,(10,10))#(start,end,shape)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.random.uniform(50,100,(3,3,3))#(start,end,shape)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
