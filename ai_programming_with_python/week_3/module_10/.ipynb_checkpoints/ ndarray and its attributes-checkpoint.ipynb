Array attributes

```python
#dimension

print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

#shape

print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

ndarray data type

```python
import numpy as np

arr =np.array([1,2,3,4])
print(arr.dtype)

arr = np.array([1.2,3.4,5.6])
print(arr.dtype)
```

```python
#upcasting

arr= np.array([1,2,3,4.5])
print(arr.dtype)

arr = np.array([1,2,3.2,'w'])
print(arr.dtype)
```

```python
arr = np.array([1,2,3],dtype=np.int8)
print(arr.dtype)

rr = np.array([1,2,300],dtype=np.int8)
print(arr)
```

```python
arr = np.array([1,2,3])
print(arr.dtype)

arr = arr.astype(np.int8)
print(arr.dtype)
```

```python
arr = np.array([1,2,3.99,'hello'],dtype=np.float64)
print(arr.dtype)
```

```python
arr = np.array([1,2,3,'hello',True])
print(arr.dtype)
```

ND ARRAY CREATION FROM EXISTING DATA

```python
lst = [10,20,30.6]

arr = np.array(lst)

print(arr)
print(type(arr))
print(arr.dtype)

lst = [10,20,30.6]

arr = np.array(lst,dtype=np.int64)

print(arr)
print(type(arr))
print(arr.dtype)

mx_lst=[10,True,'hello',10.5]


arr = np.array(mx_lst)

print(arr)
print(type(arr))
print(arr.dtype)
```

```python
matrix =[
    [1,2,3],
    [4,5,6]
]
arr = np.array(matrix)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#tuple

tup =(10,20,30)

arr = np.array(tup)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#set

st={1,2,3,4}


arr = np.array(st)
#print(arr.dtype)
#arr = np.array(st,dtype=np.int64)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
arr = np.array(list(st))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
arr = np.array(arr,dtype=np.int32)
print(arr.dtype)
```

```python
#dictionary

dc = { 'a' : 10 ,'b' : 20 ,'c' : 30}
keys = dc.keys()
print(keys)
values = dc.values()
print(values)
items=dc.items()
print(items)
```

```python
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.array(list(values))
print(arr)
arr = arr.astype(np.int8)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.array(list(items))
print(arr)
#arr = arr.astype(np.int8)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
