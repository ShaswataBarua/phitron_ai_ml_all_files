```python
import numpy as np

#np.zeros
arr = np.zeros((2,3))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr3 = np.array([
    #1st floor
    [ [1,2,3], # row1
      [4,5,6]  ], #row2

    #2nd floor

     [[1,2,3], # row1
      [4,5,6]  ], #row2


    #3rd floor
      [[1,2,3], # row1
      [4,5,6]  ] #row2


])

arr = np.zeros_like(arr3)
print(arr)
print(arr.shape)

```

```python
arr = np.ones((2,3),dtype=np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.ones_like(arr3)
print(arr)
print(arr.shape)
```

```python
arr = np.empty((2,5),dtype=np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.empty_like(arr3)
print(arr)
print(arr.shape)
```

```python
#np.full

arr = np.full((2,3),True)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.full_like(arr3,10.5)
print(arr)
print(arr.shape)
print(arr.dtype)
```

SPLITING A DATA SET

```python
import numpy as np

# dataset
data = np.array([
    [9.5, 0.5, 95.5],
    [8.2, 1.2, 86.4],
    [7.8, 1.5, 82.1],
    [7.0, 1.0, 81.0],
    [6.5, 2.0, 74.5],
    [5.0, 1.5, 70.0],
    [4.5, 3.0, 60.5],
    [3.0, 2.5, 55.0],
    [2.0, 4.0, 44.0],
    [1.5, 4.5, 39.5]
])

data
```

```python
#randomizing the data

random_index=np.random.choice(10,size=(10,),replace=False)
print(random_index)

data= data[random_index]
print(data)
```

```python
#feature and target extraction

x = data[::, :2]
y = data[::, 2:]


print(x)
print(y)
```

```python
# train , test , split

train_index = int(len(data) * 0.8) #80%

x_train = x[:train_index,:]
y_train = y[:train_index ,:]

x_test = x[train_index : ,]
y_test = y[train_index :,]

print(x_train)
print(y_train)
```
