indexing and slicing

```python
import numpy as np

#1D array

arr1 = np.array([1,2,3,4])
print(type(arr1))
print(arr1.ndim)

#2DE array

arr2 = np.array([    [1,2,3,99], #1st row
                     [4,5,6,7]   #2nd array
                ])
print(type(arr2))
print(arr2.ndim)

#3D array 


arr3= np.array([
    #1st floor
    [[1,2,3,99], 
    [4,5,6,7]],

    #2nd floor
    [[1,2,3,99],
    [4,5,6,7]]

    
])  #2D array overlap korle height add koi

print(arr3.ndim)
```

```python
print(arr1)
print(arr2)
```

```python
print(arr1[2])
```

```python
arr1[2]=999
print(arr1)
```

```python
print(arr2[0][3])
```

```python
arr2[0][3] =100
print(arr2[0][3])
```

```python
#slicing
#1D = arr[start : end : step]

arr1_mod = arr1[1 : 4]
print(arr1_mod)
arr1_mod[0]= 100000
print(arr1)


```

```python
#2D 

# arr[ row_start : row_end : step , column_start : column_end : step ]

# getiing a row

row_0 = arr2[ 0 : 1,]
row_1 = arr2[1:,]
print(row_0 )
print(row_1)

```

```python
#getting a column

col_0= arr2[:: ,0:1]
col_1 = arr2[:: ,1:2]

print(col_0)
print(col_1)
```

```python
#getting a portion

por = arr2[::,0:2]
print(por)
```
