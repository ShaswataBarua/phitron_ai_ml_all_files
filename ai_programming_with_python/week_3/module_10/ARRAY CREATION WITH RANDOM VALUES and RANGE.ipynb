ARRAY CREATION WITH RANDOM VALUES

```python
import numpy as np


arr = np.random.rand(2,3)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.random.randint(1,10,(10,10))#(start,end,shape)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr = np.random.uniform(50,100,(3,3,3))#(start,end,shape)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.arrange(first,end,step) --> similar to range

arr = np.arange(1,10,2)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

```

```python
arr = np.arange(1,10,1).reshape(3,3)
#arr = arr.reshape(3,3)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start , end(included) , koita)

arr = np.linspace(1,10,15)
#arr = arr.reshape(3,3)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

```

```python
# np.logspace(start,end(included),koita)

arr = np.logspace(0,4,6,base=2)
print(arr)
```
