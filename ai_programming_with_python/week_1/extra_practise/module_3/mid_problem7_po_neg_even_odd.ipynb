```python
inp = int(input())

n = input()
nums = n.split()
pos = 0
neg = 0
even =0
odd =0

for i in range(inp) :
    if int(nums[i]) > 0 :
        pos += 1
    if int(nums[i]) < 0 :
        neg += 1
    if int(nums[i]) % 2 == 0 :
        even += 1
    if int(nums[i]) % 2 != 0 :
        odd += 1

print("positive:" , pos)
print("negative:", neg)
print("even:", even)
print("odd", odd)
```
