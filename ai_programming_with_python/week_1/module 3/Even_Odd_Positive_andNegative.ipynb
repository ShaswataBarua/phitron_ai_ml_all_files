```python
n = int(input())
inp =input()

nums= inp.split()
even =0;odd=0;pos=0;neg=0

for i in range(n) :
    x = int(nums[i])
    if x%2 == 0 :
        even+=1
    if x%2 != 0 :
            odd += 1
    if x > 0 :
        pos+= 1
    if x < 0 :
        neg +=1


print("Even:", even)
print("Odd:", odd)
print("Positive:", pos)
print("Negative:", neg)


    
```
