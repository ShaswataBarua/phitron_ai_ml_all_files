TUPLES

```python
tup =(1,2,3,4,5)
float_tup=(23.4545,12.6767,676.2323,465.090)
mixed_tup=(10,232.454,343,"shas",True)

print(type(tup),tup)
print(type(float_tup),float_tup)
print(type(mixed_tup),mixed_tup)

not_tup=(10)
print(type(not_tup),not_tup)
tup1=(10,)
tup2=tuple([10])
print(type(tup1),tup1)
print(type(tup2),tup2)

list=[10,23,343]
tup3=tuple(list)
print(type(tup3),tup3)
```

```python
tup=(12,34,56,"str",False)

print(tup[4])

#slice

new_tup= tup[0:3]
print(new_tup)
```
