```python
e={1,2,3,4,5,6,1,2,3}

if (3 in e):
    print("3 is present")
else:
    print("3 is not present")

sum = 0

for element in e:
    sum += element

print(sum)
```

METHODS OF SET

```python
x={1,2,3,4}
y={4,5,6,7,1,2,3}

print(x.union(y))
print(x.intersection(y))
print(x.isdisjoint(y))
print(x.issubset(y))
```
