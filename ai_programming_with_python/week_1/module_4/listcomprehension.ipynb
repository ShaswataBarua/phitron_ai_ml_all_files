list comprehension

```python
even=[]

for i in range(100):
    if i%2 ==0 :
        even.append(i)

print(even)
```

```python
randomlist=[x*2 for x in range(10)]
randomlist2=[x*2 for x in range(10)]
randomlist3=[x for x in range(1,101) if x%2==0]


print(randomlist)
print(randomlist2)
print(randomlist3)
```

```python

fruit=['apple','orange','lichi']
capitalize_fruits=[]

for fruit in fruit :
    capitalize_fruits.append(fruit.capitalize())


print(capitalize_fruits)
```

```python
upper_fruit=[fruit.upper() for fruit in fruit]

print(upper_fruit)
```
