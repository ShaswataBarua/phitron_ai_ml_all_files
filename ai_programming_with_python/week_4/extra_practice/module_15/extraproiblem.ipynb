## 1. Add New Column and Save File
You have a DataFrame with column Price.
 Create a new column Discounted_Price with (10% discounted price).
 Save the modified DataFrame as discounted_products.csv.


```python
import pandas as pd 

df = pd.read_csv('Practice.csv')
df
```

```python
df['Discounted_Price'] = df['Price']  - (df['Price']* 10/100)
df
```

## 3. Handle Missing Values
Columns: Age, Salary
 Tasks:
Count missing values


Fill missing Age with mean age


Drop rows where Salary is missing



```python
print(df[['Salary','Age']].isna().sum())
```

```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
df
```

```python
df=df.dropna(subset=['Salary'])
df
```

## 4. Apply Statistical Functions
Columns: Marks1, Marks2, Marks3
 Tasks:
Calculate mean, max for each subject


Compute total marks per student


Find standard deviation of total marks



```python
print(df['Marks1'].max())
print(df['Marks2'].max())
print(df['Marks3'].max())
print(df['Marks1'].mean()) 
print(df['Marks2'].mean()) 
print(df['Marks3'].mean())
```

```python
df['marks per student'] = df['Total_Marks'] = df[['Marks1', 'Marks2', 'Marks3']].sum(axis=1) 
df
```

```python
print(df['Total_Marks'].std())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Google Drive direct download link conversion for pandas
file_id = '1Sy7rA0G9gTmeItVZ-XlugkKF4CPPqvxU'
url = f'https://drive.google.com/uc?id={file_id}'

df = pd.read_csv(url)

# Preview the first few rows and column names to confirm it loaded correctly
print(df.head())
print(df.columns)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Study_Hour'], df['IQ_Score'], color='blue', alpha=0.7)
plt.title('Study Hour vs IQ Score')
plt.xlabel('Study Hour')
plt.ylabel('IQ Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Shoe_Size'], df['Study_Hour'], color='green', alpha=0.7)
plt.title('Shoe Size vs Study Hour')
plt.xlabel('Shoe Size')
plt.ylabel('Study Hour')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Shoe_Size'], bins=10, color='orange', edgecolor='black')
plt.title('Distribution of Shoe Size')
plt.xlabel('Shoe Size')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
