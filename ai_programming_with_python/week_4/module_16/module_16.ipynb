## 16.1 Lineplot using Seaborn

## 16.8 Pairplot and Jointplot using Seaborn


```python
student_marks = student[['Marks','study_hours','attendance_rate','gender']]
student_marks
```

```python
sns.pairplot(student_marks)
```

```python
sns.pairplot(student_marks, kind ='hist')
```

```python
sns.pairplot(student_marks,hue='gender')
```

```python
# joinplot
```

```python
sns.jointplot(data=student,x='study_hours',y='Marks')
```
