```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

## 15.2 2D Line Plot Basics



```python
hours = [ 4 ,5,6,2,3,8,10]
days = [ 1,2,3,4,5,6,7]

plt.plot(days,hours)
```

```python

plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Trend of a student study time')

plt.plot(days,hours)
```

```python
hours_1 = [ 4 ,5,6,2,3,8,10]
hours_2 = [ 4 ,1,4,2,6,7,9]
days = [ 1,2,3,4,5,6,7]


plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Trend of a student study time')

plt.plot(days,hours_1)
plt.plot(days,hours_2) 
```

## 15.3 2D Line from Pandas DF


```python
df = pd.read_csv('enrollment_data.csv') 
df
```

```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment') 
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'] , df['Programming'])
```

```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment') 
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'] , df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment') 
plt.title('Student enrollment over 8 years')


plt.plot(df['Year'] , df['Programming'],label = 'Programming')
plt.plot(df['Year'], df['Digital Marketing'] , label = 'Digital Marketing')

plt.legend(loc = 'lower right')
```

## 15.4 Customizing Line plot ( color,linestyle,marker) and Grid


```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment') 
plt.title('Student enrollment over 8 years')


plt.plot(df['Year'] , df['Programming'],label = 'Programming',color='#11e8f0',linewidth=3,linestyle='dashdot',marker='o',markersize=7)
plt.plot(df['Year'], df['Digital Marketing'] , label = 'Digital Marketing',color='black',linewidth=2,linestyle='dotted')

plt.legend()
```

```python
plt.xlabel('Years')
plt.ylabel('Count of enrollment') 
plt.title('Student enrollment over 8 years')


plt.plot(df['Year'] , df['Programming'],label = 'Programming',color='#11e8f0',linewidth=3,linestyle='dashdot',marker='o',markersize=7)
plt.plot(df['Year'], df['Digital Marketing'] , label = 'Digital Marketing',color='black',linewidth=2,linestyle='dotted')

plt.legend() 

plt.grid()
plt.show()
```

## 15.6 Scatter Plots


```python
students = pd.read_csv('student_IQdata.csv') 
students
```

```python
plt.xlabel('Study hours')
plt.ylabel('IQ score') 
plt.title('Relation between the students iq and study hours')


plt.scatter( students['Study_Hour'], students['IQ_Score'])
```

```python
#relation between shoe size and iq

plt.xlabel('Shoe_Size') 
plt.ylabel('IQ_Score') 
plt.title('Shoe size vs IQ')

plt.scatter(students['Shoe_Size'] , students['IQ_Score'])
```

```python
#Relation between Chilling hour and IQ Score - > negative correlation
plt.xlabel('Chilling_Hours') 
plt.ylabel('IQ_Score') 
plt.title('Chilling_Hours vs IQ')

plt.scatter(students['Chilling_Hours'] , students['IQ_Score'],color = 'red')
```

## 15.7 Histogram


```python
plt.hist(students['IQ_Score'],color ='red', edgecolor='black') 

plt.xlabel('IQ_Score') 
plt.ylabel('Frequency of students')
```

```python
plt.hist(students['Study_Hour'],bins=15,color ='red', edgecolor='black') 

plt.xlabel('Study_Hour') 
plt.ylabel('Frequency of students')
```

```python
plt.hist(students['Study_Hour'],bins=[1,2,3,4,5,6,7,8],color ='red', edgecolor='black') 

plt.xlabel('Study_Hour') 
plt.ylabel('Frequency of students')
```

## 15.8 Bar Chart


```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
#gender wise

gender_group = df.groupby('Gender').size()
gender_group
```

```python
gender_group.index
```

```python
gender_group.values
```

```python
plt.bar(gender_group.index,gender_group.values,color = ['orange','red'],edgecolor ='black')

plt.xlabel('Gender')
plt.ylabel('Count') 
plt.title('Bar chart of gender')
```

```python
# bar chart of real data

df = pd.read_csv('student_data.csv') 
df
```

```python
status =df.groupby('CompletionStatus').size()

plt.bar(status.index,status.values,color = ['green','orange','red'],edgecolor ='black')

plt.xlabel('CompletionStatus')
plt.ylabel('Count') 
plt.title('Bar chart of CompletionStatus')
plt.show()
```

```python
#location 

status =df.groupby('Location').size()

plt.bar(status.index,status.values,edgecolor ='black')

plt.xlabel('Location')
plt.ylabel('Count') 
plt.title('Bar chart of Location')
plt.show()
```

## 15.9 Pie Chart


```python
status =df.groupby('CompletionStatus').size()

status
```

```python
plt.pie(status,labels = status.index,autopct='%1.1f%%',explode = (0.1,0,0),shadow=True)
plt.show()
```
